```python
def isValid(self, s: str) -> bool:
    stack = []
    hashmap = {")" : "(", "}" : "{", "]" : "["}

    for c in s:
        if c not in hashmap:
            stack.append(c)
            continue

        if not stack:
            return False

        popped = stack.pop()
        if popped != hashmap[c]:
            return False
    
    return not stack
```
