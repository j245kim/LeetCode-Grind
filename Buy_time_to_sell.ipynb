```python
def maxProfit(self, prices: list[int]) -> int:
    # Two pointer
    
    min_price = 100000
    max_profit = 0

    for price in prices:
        min_price = min(min_price, price)

        profit = price - min_price
        
        if profit > max_profit:
            max_profit = profit

    return max_profit
```
